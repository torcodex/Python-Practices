```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('/content/drive/MyDrive/commodity_prices.csv')
df
```

```python
df.drop(columns={'Unnamed: 0'}, inplace=True)
df.head(10)
```

```python
df.isna().sum()
```

```python
df.duplicated().sum()
```

```python
df['date'] = pd.to_datetime(df['date'])
df.info()
```

```python
df = df.sort_values(by='date')
df
```

```python
print(df['date'].min())
print(df['date'].max())
```

```python
oil_df = df[['date','oil_brent','oil_dubai']]
oil_df
```

```python
coffee_df = df[['date','coffee_arabica','coffee_robustas']]
coffee_df
```

```python
df.columns
```

```python
tea_df = df[['date','tea_columbo','tea_kolkata','tea_mombasa']]
tea_df
```

```python
recent_df = df[df['date'].dt.year >= 2015]
recent_df
```

```python
period_df = df[(df['date'].dt.year >= 2000)&(df['date'].dt.year<=2010)]
period_df
```

```python
df['month'] = df['date'].dt.month_name()
df
```

```python
monthly_oil_brent = df.groupby('month')['oil_brent'].max().sort_values(ascending=False).round(1)
monthly_oil_brent.head(10)
```

```python
df.nlargest(10,'oil_brent')[['date','oil_brent']]
```

```python
df.nsmallest(10,'sugar_world')[['date','sugar_world']]
```

```python
oil_price_gap = df['oil_brent']-df['oil_dubai']
oil_price_gap
```

```python
coffee_price_gap = df['coffee_arabica'] - df['coffee_robustas']
coffee_price_gap
```

```python
sugar_avg_df = df.groupby('month').agg(
    avg_sugar = ('sugar_world','mean'),
    avg_eu_sugar = ('sugar_eu','mean'),
    avg_us_sugar = ('sugar_us','mean')
)
sugar_avg_df
```
