```python
import pandas as pd
import numpy as np

data = {
    'Order_ID': [1001,1002,1003,1004,1005,1006,1007,1008,1009,1010,
                 1011,1012,1013,1014,1015,1015],

    'Customer': ['Rahul','Priya','Amit','Sneha','Arjun','Neha','Rohan',
                 'Anjali','Vikas','Pooja','Karan','Meera',None,'Simran',
                 'Nikhil','Nikhil'],

    'City': ['Bangalore','Mumbai','Delhi','Bangalore','Mumbai','Delhi',
             'Pune','Pune','Bangalore','Mumbai','Delhi',None,'Chennai',
             'Mumbai','Bangalore','Bangalore'],

    'Product': ['Laptop','Mobile','Headphones','Laptop','Tablet','Mobile',
                'Laptop','Headphones','Mobile','Tablet','Laptop','Mobile',
                'Headphones','Laptop',None,'Laptop'],

    'Quantity': [1,2,3,1,2,1,1,4,2,3,1,2,3,1,None,1],

    'Price': [60000,25000,2000,58000,30000,24000,62000,1800,26000,
              28000,61000,23000,2200,59000,29000,59000],

    'Discount': [5,10,0,8,5,None,10,0,5,10,12,5,0,7,10,7],

    'Payment_Mode': ['UPI','Card','UPI','Card','Cash','UPI','Card',
                     'UPI','card','Cash','UPI','Card','UPI','CARD',
                     'UPI','CARD']
}

df = pd.DataFrame(data)
```

```python
df.drop_duplicates("Order_ID", keep='last', inplace = True)
```

```python
df['Discount'] = df['Discount'].fillna(df.groupby('Product')['Discount'].transform('median'))
df['Customer'] = df['Customer'].fillna('Unknown')
df["City"] = df["City"].fillna(df["City"].mode()[0])
```

```python
df['Payment_Mode'] = df['Payment_Mode'].str.upper()
```

```python
df
```

```python
df['Gross_Amount'] = df['Quantity'] * df['Price']
df['Discount_Amount'] = df['Gross_Amount'] * (df['Discount'] / 100)
df['Final_Amount'] = df['Gross_Amount'] - df['Discount_Amount']
```
