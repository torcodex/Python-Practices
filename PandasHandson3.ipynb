```python
import pandas as pd
import numpy as np

data = { 'Order_ID': [ 501,502,503,504,505,506,507,508,509,510, 511,512,513,514,515,516,516 ],

'Customer': [
    'Rahul','Priya','Amit','Sneha','Arjun','Neha','Rohan',
    'Anjali','Vikas','Pooja','Karan','Meera',None,'Simran',
    'Nikhil','Isha','Isha'
],

'Restaurant': [
    'Burger House','Pizza Corner','Spice Hub','Burger House',
    'Pizza Corner','Food Factory','Spice Hub','Burger House',
    'Food Factory','Pizza Corner','Spice Hub','Food Factory',
    'Burger House','Pizza Corner','Food Factory','Spice Hub',
    'Spice Hub'
],

'City': [
    'Bangalore','Bangalore','Mumbai','Delhi','Mumbai','Delhi',
    'Bangalore','Mumbai','Delhi','Bangalore','Mumbai','Delhi',
    'Bangalore','Mumbai','Delhi','Bangalore','Bangalore'
],

'Order_Value': [
    450,800,650,300,1200,550,900,350,750,1100,
    500,950,400,700,1300,600,600
],

'Delivery_Time': [
    32,48,40,55,62,35,70,28,58,65,
    45,52,38,75,80,None,60
],

'Discount': [
    10,15,5,0,20,10,25,0,15,20,
    5,10,0,25,30,10,10
],

'Rating': [
    5,4,4,2,5,3,2,5,3,2,
    4,4,5,None,5,3,3
],

'Order_Status': [
    'Delivered','Delivered','Delivered','Cancelled',
    'Delivered','Delivered','Cancelled','Delivered',
    'Delivered','Cancelled','Delivered','Delivered',
    'Delivered','Cancelled','Delivered','Cancelled',
    'Cancelled'
],

'Payment_Mode': [
    'UPI','Card','UPI','Cash','Card','UPI','Card',
    'UPI','Cash','CARD','UPI','Card','UPI','card',
    'UPI','Cash','Cash'
]
}

df = pd.DataFrame(data)

df
```

```python
df.drop_duplicates("Order_ID",keep="last",inplace=True)
df
```

```python
df['Customer'].fillna("Unknown",inplace=True)
df['Rating']=df['Rating'].fillna(df.groupby("Restaurant")["Rating"].transform("median"))
df
```

```python
df['Payment_Mode'] = df['Payment_Mode'].str.upper()
df
```

```python
df['Payment_Mode'].unique()
```

```python
df['Discount_Amount'] = df["Order_Value"]*(df['Discount']/100)
```

```python
df['Final_Order_Value'] = df['Order_Value'] - df['Discount_Amount']
```

```python
def typing(a):
    if(a<=40):
        return "Fast"
    elif(a<=60):
        return 'Normal'
    else:
        return 'Slow'
df['Delivery_Performance'] = df['Delivery_Time'].apply(typing)
```

```python
df['Delivered_Orders'] = np.where(df['Order_Status']=='Delivered',1,0)

df['Cancelled_Orders'] = np.where(df['Order_Status']=='Cancelled',1,0)
```

```python
df
```

```python
Restaurant_Summary = df.groupby('Restaurant').agg(
    Total_Orders = ('Order_ID',"count"),
    Delivered_Orders = ('Delivered_Orders','sum'),
    Cancelled_Orders = ('Cancelled_Orders', 'sum'),
    Average_Order_Value = ('Order_Value','mean'),
    Average_Delivery_Time = ('Delivery_Time','mean'),
    Average_Rating = ('Rating','mean'),
    Total_Revenue = ('Final_Order_Value','sum')
)

Restaurant_Summary['Cancellation_Rate'] = (Restaurant_Summary['Cancelled_Orders']/Restaurant_Summary['Total_Orders'] *100).astype(int)
```

```python
Restaurant_Summary = Restaurant_Summary.sort_values('Cancellation_Rate',ascending=False)
```

```python
df['Discount_Group'] = np.where(df['Discount']>=20,"Discount>=20","Discount<20")
asin = df.groupby('Discount_Group')['Final_Order_Value'].mean()
if asin.iloc[0] > asin.iloc[1]:
    print(f"Average Final Order Value: {asin.iloc[0]}\nOrders with Discount<20 is more profitable")
else:
    print(f"Average Final Order Value: {asin.iloc[1]}\nOrders with Discount>=20 is more profitable")
```

```python
sums = df['Final_Order_Value'].groupby(df['Order_Status']).sum()
total_sum = sums.sum()
percent = (sums['Cancelled']/total_sum)*100
print(percent.round(1))
```
