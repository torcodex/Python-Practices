```python
combo = df.groupby('Product').agg(
    total_orders = ('Order_Status','count'),
    cancelled_orders = ('Order_Status', lambda x: x.eq('Cancelled').sum()),
    Rating=('Rating', 'mean'),
    Delivery_Time=('Delivery_Days', 'mean')
)
combo['cancellation_rate'] = ((combo['cancelled_orders']/combo['total_orders'])*100).round(1)
combo.sort_values(by='cancellation_rate', ascending=False, inplace=True)

combo['Rating'] = combo['Rating'].round(1)
combo['Delivery_Time'] = combo['Delivery_Time'].round(1)

combo['Rating_rank'] = combo['Rating'].rank(ascending=False)
combo['Delivery_Time_rank'] = combo['Delivery_Time'].rank()
combo['cancellation_rate_rank'] = combo['cancellation_rate']

combo['worst_combination_score'] = (
    combo['Rating_rank']
    + combo['Delivery_Time_rank']
    + combo['cancellation_rate_rank']
)

combo.sort_values(by='worst_combination_score', ascending=True, inplace=True)
combo.head(1)
```
