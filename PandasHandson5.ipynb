```python
df['Order_Date'] = pd.to_datetime(df['Order_Date'], format='mixed')

df['Order_Date'] = df['Order_Date'].dt.strftime('%d-%m-%Y')
df
```

```python
df['Customer_Segment'] = df['Customer_Segment'].str.title()
df['Category'] = df['Category'].str.title()
df['Payment_Method'] = df['Payment_Method'].str.upper()
```

```python
df['Shipping_Days'] = df['Shipping_Days'].fillna(df.groupby('Product')['Shipping_Days'].transform('median'))
df
```

```python
df['Rating'] = df['Rating'].fillna((df.groupby('Product')['Rating'].transform('median')).round(1))
df
```

```python
df['Order_Status'] = df['Order_Status'].str.title()
```

```python
df['Gross_Value'] = df['Quantity']*df['Unit_Price']
```

```python
df['Discount_Amount'] = df['Gross_Value'] * (df['Discount']/100)
```

```python
df['Final_Order_Value'] = df['Gross_Value'] - df['Discount_Amount']
```

```python
df['Order_Month'] = pd.to_datetime(df['Order_Date'], format='%d-%m-%Y').dt.to_period('M')
df
```

```python
conditions = [
    df['Customer_Age'] >= 56,
    (df['Customer_Age'] < 56) & (df['Customer_Age'] >= 45),
    (df['Customer_Age'] < 45) & (df['Customer_Age'] >= 35),
    (df['Customer_Age'] < 35) & (df['Customer_Age'] >= 25),
    (df['Customer_Age'] < 25) & (df['Customer_Age'] >= 18),
    df['Customer_Age'] < 18
]

values = ['Senior', 'Adult', 'Young Adult', 'Youngster', 'Teenager', 'Child']

df['Age_Group'] = np.select(conditions, values, default='Unknown')
df
```

```python
monthly_revenue_info = df.groupby('Order_Month').agg(
    total_order_value = ('Final_Order_Value','sum'),
    total_orders = ('Order_ID','nunique'),
    avg_order_value = ('Final_Order_Value','mean')
).round(1).sort_index()

monthly_revenue_info
```
